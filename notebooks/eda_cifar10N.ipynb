```python
import os
import sys
from pathlib import Path
import logging

# 1. Setup đường dẫn dự án
current_dir = Path(os.path.abspath(''))
if (current_dir / "src").exists():
    root_dir = current_dir
else:
    root_dir = current_dir.parent

if str(root_dir) not in sys.path:
    sys.path.insert(0, str(root_dir))

# 2. Import module EDA của bạn
from src.eda_analysis import perform_complete_eda

# 3. CẤU HÌNH CHO CIFAR-10N
DATA_DIR = root_dir / "notebooks" / "data_cifar10N"  # Đảm bảo thư mục này đúng với máy của bạn
# Danh sách các loại nhiễu của CIFAR-10N
NOISE_TYPES = [
    "aggre_label", 
    "worse_label", 
    "random_label1", 
    "random_label2", 
    "random_label3"
]

print("="*80)
print("BẮT ĐẦU CHẠY EDA CHO CIFAR-10N")
print("="*80)

# 4. Vòng lặp chạy EDA
for noise_type in NOISE_TYPES:
    print(f"\n🔥 Đang xử lý EDA cho loại nhiễu: {noise_type.upper()}")
    
    # Khai báo đường dẫn đến 3 file CSV
    csv_dir = DATA_DIR / "csvs" / f"noise_{noise_type}"
    paths = {
        'train_csv': str(csv_dir / "train.csv"),
        'val_csv': str(csv_dir / "val.csv"),
        'test_csv': str(csv_dir / "test.csv")
    }
    
    # Kiểm tra xem file train.csv có tồn tại không
    if not os.path.exists(paths['train_csv']):
        print(f"  ❌ Không tìm thấy data tại: {paths['train_csv']}")
        print(f"  Vui lòng chạy file prepare_cifar10n.py trước để sinh ra CSV.")
        continue
        
    # Tạo thư mục lưu biểu đồ EDA
    eda_dir = DATA_DIR / "eda_analysis" / f"noise_{noise_type}"
    eda_dir.mkdir(parents=True, exist_ok=True)
    
    try:
        # Gọi hàm EDA của bạn. Dù type hint là float, truyền chuỗi string vào vẫn chạy bình thường
        perform_complete_eda(csv_paths=paths, noise_ratio=noise_type, save_dir=str(eda_dir))
        print(f"  ✅ Thành công! Kết quả EDA đã lưu tại: {eda_dir}")
    except Exception as e:
        print(f"  ❌ Gặp lỗi khi chạy EDA cho {noise_type}: {e}")

print("\n" + "="*80)
print("🎉 ĐÃ HOÀN TẤT TOÀN BỘ QUÁ TRÌNH EDA CHO CIFAR-10N!")
print("="*80)
```
