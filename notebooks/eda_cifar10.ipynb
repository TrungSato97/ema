```python
import os
import sys
from pathlib import Path
import logging

# 1. Setup đường dẫn dự án
current_dir = Path(os.path.abspath(''))
if (current_dir / "src").exists():
    root_dir = current_dir
else:
    root_dir = current_dir.parent

if str(root_dir) not in sys.path:
    sys.path.insert(0, str(root_dir))

# 2. Import module EDA của bạn (đã được cập nhật màu mới)
from src.eda_analysis import perform_complete_eda

# =====================================================================
# 3. CẤU HÌNH CHO CIFAR-10 (Nhiễu nhân tạo)
# LƯU Ý: Hãy sửa lại DATA_DIR thành tên thư mục CIFAR-10 thật của bạn
# (Ví dụ: "data", "data_cifar10" thay vì "data_imagenet100_cmc")
# =====================================================================
DATA_DIR = root_dir / "notebooks" / "data" # <-- CHỈNH SỬA TÊN THƯ MỤC Ở ĐÂY NẾU CẦN

NOISE_RATIOS = [0.2, 0.4, 0.6, 0.8]

print("="*80)
print("BẮT ĐẦU CHẠY LẠI EDA CHO CIFAR-10 (GIAO DIỆN MÀU MỚI)")
print("="*80)

# 4. Vòng lặp chạy EDA
for noise in NOISE_RATIOS:
    print(f"\n🔥 Đang xử lý EDA cho tỷ lệ nhiễu: {noise}")
    
    # Khai báo đường dẫn đến 3 file CSV
    csv_dir = DATA_DIR / "csvs" / f"noise_{noise}"
    paths = {
        'train_csv': str(csv_dir / "train.csv"),
        'val_csv': str(csv_dir / "val.csv"),
        'test_csv': str(csv_dir / "test.csv")
    }
    
    # Kiểm tra xem file train.csv có tồn tại không
    if not os.path.exists(paths['train_csv']):
        print(f"  ❌ Không tìm thấy file CSV tại: {paths['train_csv']}")
        continue
        
    # Tạo thư mục lưu biểu đồ EDA
    eda_dir = DATA_DIR / "eda_analysis" / f"noise_{noise}"
    eda_dir.mkdir(parents=True, exist_ok=True)
    
    try:
        # Chạy EDA
        perform_complete_eda(csv_paths=paths, noise_ratio=noise, save_dir=str(eda_dir))
        print(f"  ✅ Thành công! Các biểu đồ tuyệt đẹp đã được lưu tại: {eda_dir}")
    except Exception as e:
        print(f"  ❌ Gặp lỗi khi chạy EDA cho noise {noise}: {e}")

print("\n" + "="*80)
print("🎉 ĐÃ HOÀN TẤT TOÀN BỘ QUÁ TRÌNH EDA CHO CIFAR-10!")
print("="*80)
```
