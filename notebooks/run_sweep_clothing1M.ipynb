```python


EXPERIMENT_PLAN  = {
        "alphas": [ 0.8 ,0, 0.2, 0.3, 0.5,  0.6],
        "modes": ["ema_hard"],
    }
```

```python
from pathlib import Path
import os

config = TrainConfig()
# --- BẮT BUỘC GHI ĐÈ ĐỂ AN TOÀN CHO CLOTHING1M ---
config.dataset_name = "clothing1M"
# (optional) đặt exp_name riêng cho voting experiments
config.exp_name = "clothing1M_iter_ema_real_noise"

# base dir for all runs
base_dir = Path(config.make_exp_dir())
print("BASE_DIR:", base_dir)

# config.max_iterations = 10
# config.max_epochs_per_iter = 1

for alpha in EXPERIMENT_PLAN["alphas"]:
    for mode in EXPERIMENT_PLAN["modes"]:
        # mutate config for this run

        config.alpha = float(alpha)
        config.filter_mode = mode

        # exp_dir structure (report-friendly)
        config.exp_dir = str(
            base_dir
            / f"mode_{mode}_alpha_{alpha}"
        )
        os.makedirs(config.exp_dir, exist_ok=True)

        print("\n" + "="*120)
        print("RUN:", { "alpha": alpha, "mode": mode, "exp_dir": config.exp_dir})
        print("="*120)

        run_experiment_for_setting(config, train_csv, val_csv, test_csv)
```
